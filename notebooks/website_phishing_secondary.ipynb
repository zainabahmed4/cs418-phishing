# Secondary Dataset: Website Phishing

**Source:** UCI Machine Learning Repository  
https://archive.ics.uci.edu/dataset/379/website+phishing

This notebook loads and explores the Website Phishing dataset for our CS 418 phishing research project.

```python
import arff
import polars as pl
```

```python
with open("../datasets/website-phishing/PhishingData.arff", "r") as file:
    dataset = arff.load(file)
```

```python
dataset.keys()
```

```python
# create the Polars dataframe
column_names = [attribute[0] for attribute in dataset["attributes"]]

df = pl.DataFrame(
    dataset["data"],
    schema=column_names,
    orient="row"
)

df.head()
```

## Dataset Summary

- **Rows:** 1,353
- **Columns:** 10
- **One row represents:** one website with phishing-related characteristics and a classification result
- **Missing values:** none
- **Column types:** all columns are loaded as strings
- **Target column:** `Result`
  - `-1` = phishing
  - `0` = suspicious
  - `1` = legitimate
- **Class distribution:**
  - 702 phishing websites
  - 103 suspicious websites
  - 548 legitimate websites
