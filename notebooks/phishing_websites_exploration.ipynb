## UCI Phishing Websites Dataset Exploration

```python
import arff
import polars as pl
```

```python
with open("../datasets/phishing-websites/Training Dataset.arff", "r") as file:
    dataset = arff.load(file)
```

```python
dataset.keys()
```

attributes = column names + information about each column

data       = all 11,055 rows

```python
# create the Polars dataframe
column_names = [attribute[0] for attribute in dataset["attributes"]]

df = pl.DataFrame(
    dataset["data"],
    schema=column_names,
    orient="row"
)

df.head()
```

```python
# convert the columns to integers
df = df.with_columns(
    pl.all().cast(pl.Int64)
)

df.head()
```
